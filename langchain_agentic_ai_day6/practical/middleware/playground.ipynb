```python
import subprocess

p = subprocess.Popen(
    [r"C:\WINDOWS\system32\bash.EXE"],
    stdin=subprocess.PIPE,
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True,
)

p.stdin.write("echo hello\n")
p.stdin.flush()

print(p.stdout.readline())
```
