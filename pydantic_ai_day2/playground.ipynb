```python
fruit :list[str] = ["apple", "banana", "cherry"]  # list of strings
for i in range(len(fruit)):
    print(f"Fruit {i+1}: {fruit[i]}")
```

```python
fruits :list[str] = ["apple", "banana", "cherry"]  # list of strings
fruits_a =[]

for i in fruits:
    if i.startswith("a"):
        fruits_a.append(i)

print(f"Fruits starting with 'a': {', '.join(fruits_a)}")
```
